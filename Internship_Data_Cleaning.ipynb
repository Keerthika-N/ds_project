```python
import kagglehub
path = kagglehub.dataset_download("aiexplorer77/internship-selection-prediction-dataset")
```

```python
#Locating the CSV
import os
print(os.listdir(path))
```

```python
#Loading into Pandas
import pandas as pd
file_path=os.path.join(path,"Internship_Selection_Dataset.csv")
df=pd.read_csv(file_path)
df.head()
```

```python
#Find how many selected and how many not
df['selected'].value_counts()
```

```python
#Checking if any cleaning needed
print("College Tiers:", df['college_tier'].unique())
print("Extracurriculars:", df['extracurricular'].unique())
print("Placement Training:", df['placement_training'].unique())
```

```python
#Binary Mapping
binary_mapping = {'Yes': 1, 'No':0}
df['extracurricular'] = df['extracurricular'].map(binary_mapping)
df['placement_training'] = df['placement_training'].map(binary_mapping)
df.head()
```

```python
#Tier Mapping
tier_mapping = {'Tier 1': 1, 'Tier 2': 2, 'Tier 3': 3}
df['college_tier'] = df['college_tier'].map(tier_mapping)
df.head()
```

```python
#Correlation Matrix
correlations = df.corr()['selected'].sort_values(ascending=False)
print(correlations)
```

```python
#Exporting to CSV
df.to_csv('cleaned_internship_data.csv', index=False)
```
